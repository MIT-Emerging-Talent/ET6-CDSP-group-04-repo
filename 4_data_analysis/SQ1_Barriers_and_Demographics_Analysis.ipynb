# 🎯 Introduction

This notebook explores the career pathways into the IT sector for young displaced individuals in the U.S. without prior tech backgrounds.

We address four core subquestions:
1. What % of displaced youth report cost, documentation, or language barriers to IT training?
2. What are their demographics (age, education, English proficiency, legal status)?
3. What is their employment/unemployment status?
4. Where are they geographically concentrated?

We use visual and descriptive analysis based on data from:
- Migration Policy Institute (MPI)
- UNICEF
- Upwardly Global
- ORR & ACS summaries


## 📌 Barrier Overview: Cost, Documentation, Language

Key barriers facing displaced youth:

- **Cost**: Significant affordability challenges; nearly 50% of refugee jobseekers lost jobs early in COVID, 69% had <1 month’s savings.
- **Documentation**: Credential recognition, lack of work permits, and asylum backlogs delay access to jobs or training.
- **Language**: 47% of immigrants age 5+ speak English less than “very well”; even higher for displaced populations.

UNICEF and Upwardly Global confirm these are the most systemic barriers to IT pathways.


## 👥 Demographics of Displaced Youth

- **Age**: Most immigrants are 18–64 (77%), likely capturing youth.
- **Legal Status**: 73% of immigrants have legal status; 13.7M are undocumented.
- **Education**: 48% of recent arrivals have college degrees, but many are underemployed.
- **English**: 46% are Limited English Proficient (LEP).
- **Origins**: Displaced youth often come from Syria, DRC, Afghanistan, Venezuela, etc.

🧠 Insight: 31% of immigrants arrived post-2010, indicating many are likely youth or young adults.


## 💼 Employment Patterns

- Immigrants = 17% of U.S. labor force (28.6M people).
- Many displaced youth are unemployed or underemployed due to delays in work authorization.
- Credential mismatch and employer bias also block access to IT jobs.
- 2M+ skilled immigrants are underutilized in U.S. job market.

🧠 Upwardly Global data confirms refugee youth are eager to work and upskill, but face structural barriers.


## 🗺️ Geographic Concentration

Displaced youth are most likely concentrated in:

- California (10.4M immigrants)
- Texas (5.2M)
- Florida (4.8M)
- New York (4.5M)

Urban centers like NYC, LA, Houston, Miami, and Chicago serve as integration hubs with existing services and training opportunities.


## 🎯 Program Design Insights (UNICEF & Upwardly Global)

Best practices for supporting displaced youth in IT include:

- **Micro-credentials**: stackable, job-relevant, flexible
- **Mobile-first learning**: low-cost and scalable
- **Mentorship & peer learning**: boosts confidence and language skills
- **Credential translation support**: to overcome documentation barriers
- **Localized content**: tailored to culture/language needs

These strategies directly map to reducing the 3 main barriers: cost, documentation, and language.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("../1_datasets/cleaned_data/master_displaced_youth_it_pathways.csv")
df.head()
```

### 🧱 Top 10 States – Cost Barrier

```python
barrier_top10 = df.sort_values("% Citing Cost Barrier", ascending=False).head(10)
plt.figure(figsize=(10, 5))
sns.barplot(data=barrier_top10, x="% Citing Cost Barrier", y="Region or State")
plt.title("Top 10 States – Cost Barrier")
plt.tight_layout()
plt.show()
```

### 🧱 Top 10 States – Documentation Barrier

```python
barrier_top10 = df.sort_values("% Citing Documentation Barrier", ascending=False).head(
    10
)
plt.figure(figsize=(10, 5))
sns.barplot(data=barrier_top10, x="% Citing Documentation Barrier", y="Region or State")
plt.title("Top 10 States – Documentation Barrier")
plt.tight_layout()
plt.show()
```

### 🧱 Top 10 States – Language Barrier

```python
barrier_top10 = df.sort_values("% Citing Language Barrier", ascending=False).head(10)
plt.figure(figsize=(10, 5))
sns.barplot(data=barrier_top10, x="% Citing Language Barrier", y="Region or State")
plt.title("Top 10 States – Language Barrier")
plt.tight_layout()
plt.show()
```

### 🌍 Top 5 States by Displaced Youth Population

```python
top_states = df.sort_values(
    "Estimated Displaced Youth Population (18–35)", ascending=False
).head(5)
plt.figure(figsize=(8, 5))
sns.barplot(
    data=top_states,
    x="Estimated Displaced Youth Population (18–35)",
    y="Region or State",
    palette="viridis",
)
plt.title("Top 5 States by Displaced Youth Population")
plt.tight_layout()
plt.show()
```

### 📋 Count of Regions by Legal Status Group

```python
plt.figure(figsize=(7, 5))
sns.countplot(
    data=df,
    y="Dominant Immigration/Legal Status Group",
    order=df["Dominant Immigration/Legal Status Group"].value_counts().index,
    palette="pastel",
)
plt.title("Count of Regions by Dominant Legal Status Group")
plt.tight_layout()
plt.show()
```

### 📈 Education vs Unemployment – All Groups

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="% with Post-Secondary Education",
    y="% Unemployed",
    hue="Dominant Immigration/Legal Status Group",
    size="Estimated Displaced Youth Population (18–35)",
    sizes=(40, 250),
    alpha=0.7,
)
plt.title("Education vs. Unemployment Among Displaced Youth")
plt.xlabel("% with Post-Secondary Education")
plt.ylabel("% Unemployed")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

### 🔎 Refugee – Education vs. Unemployment (Labeled)

```python
subset = df[df["Dominant Immigration/Legal Status Group"] == "Refugee"]
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=subset,
    x="% with Post-Secondary Education",
    y="% Unemployed",
    size="Estimated Displaced Youth Population (18–35)",
    sizes=(50, 300),
    color="steelblue",
    alpha=0.7,
    legend=False,
)
for i, row in subset.iterrows():
    plt.text(
        row["% with Post-Secondary Education"] + 0.3,
        row["% Unemployed"],
        row["Region or State"],
        fontsize=9,
    )
plt.title("Refugee: Education vs. Unemployment with State Labels")
plt.xlabel("% with Post-Secondary Education")
plt.ylabel("% Unemployed")
plt.grid(True)
plt.tight_layout()
plt.show()
```
